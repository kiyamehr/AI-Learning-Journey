```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest
```

```python
data = pd.read_csv('servers_data.csv')

x = data.values

print(x.shape)
print(data.head())
print(data.tail())
```

```python
model = IsolationForest(contamination=0.1, random_state=36)

model.fit(x)
yhat=  model.predict(x)

print(yhat)
```
